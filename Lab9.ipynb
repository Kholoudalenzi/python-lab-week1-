```python
import numpy as np

A = np.arange(1, 21).reshape(4, 5)
print("A=\n", A)
print("A dtype:", A.dtype)
print("A ndim:", A.ndim)
print("A shape:", A.shape)
print("A size:", A.size)
print("A itemsize:", A.itemsize)

B = A.reshape(5, 4)
print("\nB=\n", B)
print("B.shape:",B.shape)
```

```python
import numpy as np
a = np.array([[0., 0., 0.],
              [10., 10., 10.],
              [20., 20., 20.],
              [30., 30., 30.]])
b = np.array([0., 1., 2.])

c = a + b
print("c=\n", c)

mask = (c >= 22)
print("\nMask=\n", mask)

count = mask.sum()
print("\nCount of values >= 22:", count)
```

```python
grades = np.array([[87, 96, 70],
                   [100, 87, 90],
                   [94, 77, 90],
                   [100, 81, 82]])

overall_sum = np.sum(grades)
overall_min = np.min(grades)
overall_max = np.max(grades)
overall_mean = np.mean(grades)
overall_std = np.std(grades)
overall_var = np.var(grades)

print("Overall sum:", overall_sum)
print("Overall min:", overall_min)
print("Overall max:", overall_max)
print("Overall mean:", overall_mean)
print("Overall std:", overall_std)
print("Overall var:", overall_var)

mean_per_col = np.mean(grades, axis=0)
mean_per_row = np.mean(grades, axis=1)

print("Mean per column:", mean_per_col)
print("Mean per row:", mean_per_row)

max_per_row = np.max(grades, axis=1)
print("Max per row:", max_per_row)
```

```python
import numpy as np
matrix = np.array([[2, 4, 6],
                   [1, 3, 5],
                   [7, 9, 11]])

sum_per_row = np.sum(matrix, axis=1)
sum_per_col = np.sum(matrix, axis=0)
overall_mean = np.mean(matrix)

print("Sum per row:", sum_per_row)
print("Sum per column:", sum_per_col)
print("Overall mean:", overall_mean)
```

```python
import numpy as np
grades = np.array([[70, 85, 90],
                   [88, 76, 95],
                   [60, 80, 72]])

max_index = np.unravel_index(np.argmax(grades), grades.shape)
print("Index of max value:", max_index)

max_per_row_index = np.argmax(grades, axis=1)
print("Index of max per row:", max_per_row_index)

min_per_col_index = np.argmin(grades, axis=0)
print("Index of min per column:", min_per_col_index)
```

```python
import numpy as np
def manhattan_distance(x1, y1, x2, y2):
    return abs(x2 - x1) + abs(y2 - y1)

def euclidean_distance(x1, y1, x2, y2):
    return np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# ex
p1, p2 = (1, 2), (4, 6)
print("Manhattan Distance:", manhattan_distance(p1[0], p1[1], p2[0], p2[1]))
print("Euclidean Distance:", euclidean_distance(p1[0], p1[1], p2[0], p2[1]))
```

```python
try:
    n1 = float(input("enter first number: "))
    n2 = float(input("enter second number: "))

    ans = n1 / n2

except ZeroDivisionError:
    print("cannot divide by zero")

except ValueError:
    print("please enter numbers only")

else:
    print("result:", ans)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age
    def set_age(self, new_age):
         self.__age = new_age
    def display(self):
        print("Name =", self.name)
        print("Age =", self.__age)

p = Person("Kholoud", 20)

p.display()

p.set_age(21)

print("After updating age :")
p.display()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):
    def add_bonus(self, bonus):
        self.salary += bonus

    def display(self):
        print("name:", self.name)
        print("total salary:", self.salary)


manager1 = Manager("sara", 4500)

manager1.add_bonus(500)

manager1.display()
```

```python
class Student:
    def __init__(self):
        self.name = input("enter name: ")

        try:
            self.age = int(input("enter age: "))
            if self.age < 0:
                print("age cannot be negative")
        except ValueError:
            print("invalid age")

    def display(self):
        print("name:", self.name)
        print("age:", self.age)


student1 = Student()
student1.display()
```

```python
import os

os.makedirs("lab4_data/input", exist_ok=True)
os.makedirs("lab4_data/output", exist_ok=True)
os.makedirs("lab4_data/logs", exist_ok=True)

for item in os.listdir("lab4_data"):
    print(item)

os.rename("lab4_data/logs", "lab4_data/history")
```

```python
with open("lab4_data/input/notes.txt", "w") as f:
    f.write("Python is easy\n")
    f.write("Files are useful\n")
    f.write("This is Lab 4\n")

with open("lab4_data/input/notes.txt", "r") as f:
    lines = f.readlines()
    print("Lines:", len(lines))

with open("lab4_data/input/notes.txt", "r") as f:
    text = f.read()

with open("lab4_data/input/summary.txt", "w") as f:
    f.write(text)
```

```python
import pyfiglet

ascii_text = pyfiglet.figlet_format("KHOLOUD")
print(ascii_text)
os.makedirs("lab4_data/output", exist_ok=True)
output_file = "lab4_data/output/ascii_name.txt"
with open (output_file, "w") as f:
  f.write(ascii_text)
print(f"ASCCII text saved to{output_file}")
```
